```python
# Codes for duplicate
from collections import Counter

string = input("Enter a string: ")

count = Counter(string.replace(" ", ""))

print("Duplicate characters are:")

for char, freq in count.items():
    if freq > 1:
        print(char, ":", freq)
```

```python
# Codes for reverse
string = input("Enter a string: ")

reverse = string[::-1]
print("Reversed string:", reverse)
```

```python
a,b,c,d=10,20,10,20
print(id(a))
print(id(b))
print(id(c))
print(id(d))
```

Why are the addresses the same?

a and c both store the value 10.

b and d both store the value 20.

Python reuses (interns/caches) commonly used immutable objects such as small integers.

Instead of creating a new object for every 10 or 20, Python stores one object in memory and lets multiple variables point to it.

Therefore:
a and c reference the same object → same id().

b and d reference the same object → same id().

# Python Data Types

A data type defines the type of value a variable can store. Python automatically detects the data type when you assign a value to a variable (dynamic typing).

**Definition**:

• A data type specifies the kind of data stored in a variable.

• Python is dynamically typed, so you do not need to declare the data type explicitly.

• You can check a variable's data type using the type() function.


**1. Number** : Number Data Types store numerical values and they are immutable data types, means that changing values of number data type results in a newly allocated object.

**2. Integer** : An integer (int) is a whole number without any decimal point. It can be positive, negative, or zero.

**3. Floating-Point** :A floating-point number (float) is a numeric data type that represents numbers with a decimal point or numbers written in scientific (exponential) notation. It is used to store fractional or real values.

Representation:

 Written with a decimal point (e.g., 10.5, 3.14, -0.75).
	￼	Can also be written in scientific notation using e or E (e.g., 2.5e3 = 2500.0, 4.2E-2 = 0.042).

**4. Complex Number**: A complex number (complex) is a numeric data type that represents numbers with both a real part and an imaginary part. In Python, the imaginary part is written using j.

Representation:
	￼	Written in the form a + bj, where a is the real part and b is the imaginary part.

```python
# Integer Example

a = 10
b = 5

print("Value of a =", a)
print("Value of b =", b)

print("Data type of a =", type(a))
print("Data type of b =", type(b))

print("Addition =", a + b)
print("Subtraction =", a - b)
print("Multiplication =", a * b)
print("Division =", a / b)
```

```python
# Complex Number Example

a = 3 + 4j
b = 1 + 2j

print("Value of a =", a)
print("Value of b =", b)

print("Data type of a =", type(a))
print("Data type of b =", type(b))

print("Addition =", a + b)
print("Subtraction =", a - b)
print("Multiplication =", a * b)
```
